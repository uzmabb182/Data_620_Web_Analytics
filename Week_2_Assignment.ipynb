```python
# importing NetworkX for the graph math and Matplotlib
import matplotlib.pyplot as plt
import networkx as nx
print(nx.__version__)
```

```python
# Initialize the Graph by creating a blank container to hold your network data using undirected graph, meaning the relationships are mutual.

G = nx.Graph()
```

```python
# Adding specified nodes
# The individuals are the nodes, and the working relationships between them are the edges.
G.add_nodes_from(["Max", "Alex", "Farry", "Jackie"])

# Connect the nodes to form relationships
G.add_edges_from([
    ("Max", "Alex"),
    ("Max", "Farry"),
    ("Max", "Jackie"),
    ("Alex", "Jackie"),
    ("Farry", "Jackie")
])
```

```python
# Print the nodes and edges to verify
print("Team Members (Nodes):", list(G.nodes()))
print("Collaborations (Edges):", list(G.edges()))
```

```python
# Set the canvas size
plt.figure(figsize=(8, 6))

# Drawing the network
nx.draw(G,
        with_labels=True,
        node_color='skyblue',
        edge_color='gray',
        node_size=3000,
        font_size=12,
        font_weight='bold')

# Adding a title and displaying the plot
plt.title("Hello Graph World")
plt.show()
```
