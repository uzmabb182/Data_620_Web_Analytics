# Lexical Diversity

Lexical diversity is a mathematical measure of how rich and varied a text's vocabulary is, calculated by dividing the number of unique words by the total number of words.

The table compares vocabulary usage across six genres in the Brown Corpus.
- News has the most tokens (100,554) and unique words (14,394), indicating the largest text collection.

- Romance has the highest lexical diversity score (8.3), indicating greater word repetition relative to its vocabulary size.

- Humor has the lowest score (4.3), suggesting less repetition and greater vocabulary variety relative to its length.

- Science fiction has the fewest tokens (14,470) and unique words (3,233).
- Hobbies and religion fall between these extremes.

```python
# Select genres from the Brown Corpus
genres = [
    'hobbies',
    'humor',
    'science_fiction',
    'news',
    'romance',
    'religion'
]

# Create an empty list
results = []

# Calculate statistics for each genre
for genre in genres:
    words = brown.words(categories=genre)

    tokens = len(words)
    types = len(set(words))
    lexical_diversity = round(tokens / types, 1)

    results.append([
        genre,
        tokens,
        types,
        lexical_diversity
    ])

# Convert results into a DataFrame
df = pd.DataFrame(
    results,
    columns=['Genre', 'Tokens', 'Types', 'Lexical Diversity']
)

display(df)
```

# cumulative percentage

The cumulative percentage tells us what percentage of all tokens in Moby Dick is represented by the most frequent words combined.

Cumulative %=(Cumulative Frequency*Total Tokens)100

For the first token:
{18,713}{260,819}*100=7.17%

For the first two tokens combined:

{18,713+13,721}{260,819}*100=12.44%


Notice that we keep adding the frequencies, which is why the cumulative percentage always increases.


The first token accounts for approximately 7% of the text.



The first five tokens account for approximately 20%.



The 50 most frequent tokens account for nearly 47%.

Key takeaway: A very small number of frequently occurring tokens make up almost half of Moby Dick. This illustrates how word frequencies are highly uneven in natural language.

# Define Functions and Variables
Create a custom tool **to measure lexical diversity** and list the genres to analyze.

```python
def lexical_diversity(text):
    return len(set(text)) / len(text)

genre_stats = {}
genres = ['news', 'romance', 'humor']
```

# Iterate, Calculate, and Store

Loop through each genre, calculate the total tokens and diversity, and store them in the genre_stats dictionary.

```python
for genre in genres:
    genre_words = nltk.corpus.brown.words(categories=genre)
    genre_stats[genre] = {
        'total_tokens': len(genre_words),
        'diversity': lexical_diversity(genre_words)
    }

# Peek inside the dictionary for the 'humor' category
print(genre_stats['humor'])
```

Loop through the finished dictionary to print the data in a clean, readable format. The :.3f rounds the diversity score to three decimal places.

This output demonstrates that the "humor" genre features the richest vocabulary among the three categories, despite having the smallest overall word count. The Python loop extracts and formats the stored dictionary data to give you a clear comparison of text length versus vocabulary variety.

News: With 100,554 total tokens, it has the largest dataset but a moderate lexical diversity score of 0.143. This means about 14.3% of the words used are unique.

Romance: This category contains 70,022 total words but yields the lowest diversity score at 0.121, indicating writers in this genre rely more heavily on repeating the same core vocabulary.

Humor: Although it provides the smallest text sample at just 21,695 tokens, it achieves the highest diversity score of 0.231. Nearly a quarter of its words are unique, showing a much wider range of word choices.

```python
for category, stats in genre_stats.items():
    print(f"Genre: {category} | Tokens: {stats['total_tokens']} | Diversity: {stats['diversity']:.3f}")
```
